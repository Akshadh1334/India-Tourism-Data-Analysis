```python
import pandas as pd
import matplotlib.pyplot as plt

df.columns = df.columns.str.strip()

numeric_cols = [
    "Arrivals (In Numbers)",
    "North (In Percentage)",
    "South (In Percentage)",
    "West and Central (In Percentage)",
    "East (In Percentage)",
    "North East (In Percentage)"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=["Country/Territory", "Arrivals (In Numbers)"])

print("Dataset rows and columns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst five rows:")
display(df.head())
```

```python

top10 = df.nlargest(10, "Arrivals (In Numbers)").sort_values(
    "Arrivals (In Numbers)", ascending=True
)

plt.figure(figsize=(12, 7))
plt.barh(top10["Country/Territory"], top10["Arrivals (In Numbers)"])

plt.title("Top 10 Entries by Tourist Arrivals in India (2024)")
plt.xlabel("Number of Arrivals")
plt.ylabel("Country / Territory")
plt.grid(axis="x", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

```python

print(df["Country/Territory"].to_list())
```

```python
selected_country = "United States"

regions = {
    "North India": "North (In Percentage)",
    "South India": "South (In Percentage)",
    "West and Central India": "West and Central (In Percentage)",
    "East India": "East (In Percentage)",
    "North East India": "North East (In Percentage)"
}

selected_row = df[
    df["Country/Territory"].astype(str).str.strip().str.casefold()
    == selected_country.strip().casefold()
]

if selected_row.empty:
    print("Entry not found. Check the exact name in the list above.")
else:
    values = selected_row.iloc[0][list(regions.values())].fillna(0)
    values.index = list(regions.keys())

    values = values[values > 0]

    plt.figure(figsize=(8, 8))
    plt.pie(
        values,
        labels=values.index,
        autopct="%1.1f%%",
        startangle=90
    )
    plt.title(f"Regional Distribution for {selected_country}")
    plt.tight_layout()
    plt.show()
```

```python
region_columns = {
    "North India": "North (In Percentage)",
    "South India": "South (In Percentage)",
    "West and Central India": "West and Central (In Percentage)",
    "East India": "East (In Percentage)",
    "North East India": "North East (In Percentage)"
}

regional_average = df[list(region_columns.values())].mean()
regional_average.index = list(region_columns.keys())


regional_average = regional_average.sort_values(ascending=False)

plt.figure(figsize=(11, 6))
plt.bar(regional_average.index, regional_average.values)

plt.title("Average Regional Distribution Across Dataset Entries")
plt.xlabel("Region of India")
plt.ylabel("Average Percentage (%)")
plt.xticks(rotation=25, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()

print("Average regional percentages:")
display(regional_average.round(2).to_frame("Average Percentage"))
```

```python
region_cols = list(region_columns.values())
print("STATISTICAL SUMMARY")
display(df[["Arrivals (In Numbers)"] + region_cols].describe().round(2))
print("\nAVERAGE REGIONAL PERCENTAGES")
display(
    df[region_cols].mean()
    .rename(index={
        "North (In Percentage)": "North India",
        "South (In Percentage)": "South India",
        "West and Central (In Percentage)": "West and Central India",
        "East (In Percentage)": "East India",
        "North East (In Percentage)": "North East India"
    })
    .sort_values(ascending=False)
    .round(2)
    .to_frame("Average Percentage")
)

df["Most Prominent Region"] = (
    df[region_cols].idxmax(axis=1)
    .str.replace(" (In Percentage)", "", regex=False)
)
print("\nMOST PROMINENT REGION BY ENTRY")
print("\nNUMBER OF ENTRIES LED BY EACH REGION")
display(
    df["Most Prominent Region"]
    .value_counts()
    .to_frame("Number of Entries")
)
```

```python
correlation = df[
    ["Arrivals (In Numbers)"] + region_cols
].corr(numeric_only=True)

plt.figure(figsize=(10, 7))
plt.imshow(correlation, aspect="auto")
plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=70,
    ha="right"
)
plt.yticks(range(len(correlation.index)), correlation.index)

plt.title("Correlation Between Arrivals and Regional Percentages")
plt.tight_layout()
plt.show()

display(correlation.round(2))
```

```python

top_entry = df.loc[df["Arrivals (In Numbers)"].idxmax()]

average_regions = df[region_cols].mean()
best_region_column = average_regions.idxmax()

region_names = {
    "North (In Percentage)": "North India",
    "South (In Percentage)": "South India",
    "West and Central (In Percentage)": "West and Central India",
    "East (In Percentage)": "East India",
    "North East (In Percentage)": "North East India"
}

print("=" * 55)
print("TOURISM DATA ANALYSIS: KEY FINDINGS")
print("=" * 55)

print(
    f"\n1. Highest-arrival dataset entry: "
    f"{top_entry['Country/Territory']}"
)
print(
    f"   Recorded arrivals: "
    f"{top_entry['Arrivals (In Numbers)']:,.0f}"
)

print(
    f"\n2. Highest average regional percentage: "
    f"{region_names[best_region_column]}"
)
print(
    f"   Average across dataset entries: "
    f"{average_regions[best_region_column]:.2f}%"
)

print("\n3. Average percentage by region:")
for column, value in average_regions.sort_values(ascending=False).items():
    print(f"   {region_names[column]}: {value:.2f}%")

print("\n4. Most frequently leading region:")
print(
    df["Most Prominent Region"].value_counts().idxmax()
    if "Most Prominent Region" in df.columns
    else df[region_cols].idxmax(axis=1).mode().iloc[0]
)

print("\n" + "=" * 55)
print("CONCLUSION")
print("=" * 55)

print(
    "The analysis compares tourist arrivals and regional distribution "
    "percentages across entries in the dataset. The visualisations and "
    "statistical summaries help identify differences in recorded "
    "regional distributions. Because the dataset includes aggregate "
    "groups as well as individual countries, the results should be "
    "interpreted with that limitation in mind."
)
```

```python
region_columns = {
    "North India": "North (In Percentage)",
    "South India": "South (In Percentage)",
    "West and Central India": "West and Central (In Percentage)",
    "East India": "East (In Percentage)",
    "North East India": "North East (In Percentage)"
}

plot_df = df.copy()
plot_df["Arrivals (In Numbers)"] = pd.to_numeric(
    plot_df["Arrivals (In Numbers)"], errors="coerce"
)

plt.figure(figsize=(11, 7))

for region_name, column in region_columns.items():
    plot_df[column] = pd.to_numeric(plot_df[column], errors="coerce")

    valid = plot_df[
        (plot_df["Arrivals (In Numbers)"] > 0)
        & plot_df[column].notna()
    ]

    plt.scatter(
        valid["Arrivals (In Numbers)"],
        valid[column],
        alpha=0.55,
        label=region_name
    )

plt.title("Tourist Arrivals vs Regional Distribution")
plt.xlabel("Number of Tourist Arrivals")
plt.ylabel("Regional Percentage (%)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

region_cols = [
    "North (In Percentage)",
    "South (In Percentage)",
    "West and Central (In Percentage)",
    "East (In Percentage)",
    "North East (In Percentage)"
]

heatmap_df = df[["Arrivals (In Numbers)"] + region_cols].apply(
    pd.to_numeric, errors="coerce"
)

correlation = heatmap_df.corr()

plt.figure(figsize=(10, 7))
plt.imshow(correlation, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Correlation Coefficient")

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=70,
    ha="right"
)
plt.yticks(range(len(correlation.index)), correlation.index)

for i in range(len(correlation.index)):
    for j in range(len(correlation.columns)):
        value = correlation.iloc[i, j]
        if pd.notna(value):
            plt.text(j, i, f"{value:.2f}", ha="center", va="center")

plt.title("Correlation Heatmap of Tourism Data")
plt.tight_layout()
plt.show()
```

```python
region_names = {
    "North (In Percentage)": "North India",
    "South (In Percentage)": "South India",
    "West and Central (In Percentage)": "West and Central India",
    "East (In Percentage)": "East India",
    "North East (In Percentage)": "North East India"
}

regional_data = df[region_names.keys()].apply(
    pd.to_numeric, errors="coerce"
)

df["Most Prominent Region"] = (
    regional_data.idxmax(axis=1).map(region_names)
)

region_counts = df["Most Prominent Region"].value_counts()

plt.figure(figsize=(10, 6))
plt.bar(region_counts.index, region_counts.values)

plt.title("Most Prominent Region Across Dataset Entries")
plt.xlabel("Region")
plt.ylabel("Number of Entries")
plt.xticks(rotation=25, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()

display(region_counts.to_frame("Number of Entries"))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Prepare numeric data
scatter_df = df[
    ["Arrivals (In Numbers)", "South (In Percentage)"]
].copy()

scatter_df["Arrivals (In Numbers)"] = pd.to_numeric(
    scatter_df["Arrivals (In Numbers)"], errors="coerce"
)

scatter_df["South (In Percentage)"] = pd.to_numeric(
    scatter_df["South (In Percentage)"], errors="coerce"
)

scatter_df = scatter_df.dropna()
scatter_df = scatter_df[scatter_df["Arrivals (In Numbers)"] > 0]

# Draw scatter plot
plt.figure(figsize=(10, 6))

plt.scatter(
    scatter_df["Arrivals (In Numbers)"],
    scatter_df["South (In Percentage)"],
    alpha=0.6
)

plt.title("Tourist Arrivals vs South India Percentage")
plt.xlabel("Number of Tourist Arrivals")
plt.ylabel("South India Percentage (%)")
plt.grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
summary = pd.DataFrame({
    "Region": list(region_columns.keys()),
    "Mean Percentage": [
        pd.to_numeric(df[col], errors="coerce").mean()
        for col in region_columns.values()
    ],
    "Median Percentage": [
        pd.to_numeric(df[col], errors="coerce").median()
        for col in region_columns.values()
    ],
    "Minimum Percentage": [
        pd.to_numeric(df[col], errors="coerce").min()
        for col in region_columns.values()
    ],
    "Maximum Percentage": [
        pd.to_numeric(df[col], errors="coerce").max()
        for col in region_columns.values()
    ]
})

summary = summary.round(2)
display(summary)

summary.to_csv("tourism_regional_analysis.csv", index=False)

print("Regional analysis saved successfully!")
```
